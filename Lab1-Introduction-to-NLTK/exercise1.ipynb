# Exercise 1: Tokenization and Stop Words using NLTK

This notebook loads the text from `myself.txt`, tokenizes it, identifies the stop words present, and prints the sentences with stop words removed using Python's NLTK library.

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Download the required NLTK resources
nltk.download("punkt_tab", quiet=True)
nltk.download("stopwords", quiet=True)

# A set of sentences about myself
text_about_me = """
My name is Sujith.
I enjoy building software projects.
I like exploring new technologies.
I spend my free time learning programming.
My goal is to create useful applications.
"""

# Tokenize the text
raw_tokens = word_tokenize(text_about_me)

# Remove punctuation and retain only words
tokens = [token for token in raw_tokens if token.isalpha()]

# Load the English stop-word collection
english_stop_words = set(stopwords.words("english"))

# Find the stop words that occur in the given text
found_stop_words = [
    token for token in tokens
    if token.lower() in english_stop_words
]

# Remove stop words from the tokens
filtered_tokens = [
    token for token in tokens
    if token.lower() not in english_stop_words
]

# Print the results
print("TEXT ABOUT ME:")
print(text_about_me)

print(f"\nTOTAL NUMBER OF TOKENS: {len(tokens)}")

print("\nTOKENS:")
print(tokens)

print(f"\nTOTAL STOP-WORD OCCURRENCES: {len(found_stop_words)}")

print("\nSTOP WORDS FOUND IN THE TEXT:")
print(found_stop_words)

print("\nUNIQUE STOP WORDS FOUND:")
print(sorted({word.lower() for word in found_stop_words}))

print("\nTOKENS AFTER REMOVING STOP WORDS:")
print(filtered_tokens)
```
